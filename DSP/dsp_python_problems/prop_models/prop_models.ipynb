### Propagation model comparison

```python
import prop_models as pm
import numpy as np
import matplotlib.pyplot as plt
```

### Simulation parameters

```python
distance_vector = np.linspace(0.001,0.1,40)

station_height = 10
ue_height = 1.5
freq_ghz = 2
```

### Computations

```python
fspl = np.arange(0,len(distance_vector))
ohata = np.arange(0,len(distance_vector))
itur = np.arange(0,len(distance_vector))
sui = np.arange(0,len(distance_vector))

for index,d in enumerate(distance_vector):
    fspl[index] = pm.free_space_loss(d, freq_ghz)
    ohata[index] = pm.okumura_hata_loss(d,freq_ghz*1e3,station_height,ue_height)
    itur[index] = pm.itu_r_p1238(d*1e3,freq_ghz,0,'residential')
    sui[index] = pm.sui_model(d*1e3, freq_ghz * 1e3, station_height,ue_height)
```

### Plotting

```python
plt.plot(1e3*distance_vector,ohata,'o')
plt.plot(1e3*distance_vector,fspl,'o')
plt.plot(1e3*distance_vector,itur,'o')
plt.plot(1e3*distance_vector,sui,'o')

plt.xlabel('distance [m]')
plt.ylabel('Amplitude Loss [dB]')

plt.legend(['okumura-hata','FSPL', 'ITU-R P 1238', 'SUI'])
plt.ylim(0,150)
plt.show()
```
